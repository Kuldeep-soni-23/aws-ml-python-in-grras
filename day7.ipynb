```python
class User1:
  def __init__(self,pen,pencil):
    self.pen = pen
    self.pencil = pencil
  def penGet(self):
    print("pen is working: ",self.pen)
  def pencilGet(self):
    print("pencil is working: ",self.pencil)
 
class User2(User1):
  def __init__(self,pen,pencil,earser):
    self.pen = pen
    self.pencil = pencil
    self.earser = earser
# object
u = User2("blue","black","doms")
u.penGet()
 
```

```python
class Kuldeep:
    def __init__(self,bike):
        self.bike = bike
    def usebike(self):
        print("Bike used by: ",self.bike)
class Mandeep(Kuldeep):
    def __init__(self, name,bike):
        self.name = name
        super().__init__(bike)
    def usename(self):
        print("name is: ",self.name)

u = Mandeep("mandeep friend","tvs by mandeep")
u.usename()
u.usebike()
```

```python
# encapsulation
class Student:
  def __init__(self,name,roll_no):
    self.__data = name # name is attribute
    self.roll_no = roll_no # roll_no is attribute
  def getname(self,password=None):
    if password == "123456":
      print(self.__data)
    else:
      print("password not match")
 
s = Student("rajendra",123)
# print(s.__data) # no attribute
print(s.roll_no)
s.getname("123456")
 
# example
 
```

```python
#overriding
class User1:
  def __init__(self,pen,pencil):
    self.pen = pen
    self.pencil = pencil
  def getpen(self):
    print("user1: ",self.pen)
 
class User2(User1):
  def __init__(self,pen,pencil,earser):
    self.earser = earser
    super().__init__(pen,pencil)
  def getpen(self):
    print("user2: ",self.pen)
 
u = User2("blue","black","doms")
u.getpen()
 
```
