```python
# while question
print("loop start")
while True:
  num = int(input('Enter a number: '))
  if num == 10:
    print(f"number {num} is found")
    break
 
print("loop end")
 
```

```python
def cal(a, b, op):
    if op == "+":
        print(a + b)
    elif op == "-":
        print(a - b)
    elif op == "*":
        print(a * b)
    elif op == "/":
        print(a / b)
    elif op == "%":
        print(a % b)
    elif op == "//":
        print(a // b)
    else:
        print("Error")

cal(10, 20, "+")
```

```python
def cal(a, b, op):
    print(eval(f"{a}{op}{b}"))

cal(10, 20, "-")
cal(10, 20, "+")
cal(10, 20, "*")
cal(10, 20, "/")
cal(10, 20, "%")
cal(10, 20, "//")
```

```python
try:
    print(10/0)
    # print("hello" + 10)

except TypeError:
    print("type error")

except ZeroDivisionError:
    print("zero division error")
```

```python
try:
    print("try part")
    print(10/0)
except:
    print("except part")
finally:
    print("finally part")
```
