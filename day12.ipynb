```python
import pandas as pd

data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Riya", "Raj", "Sara", "John", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000, 52000],
    "Experience": [2, 3, 5, 4, 1, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
#filter one greater
df[df['Emp_ID']>104]
```

```python
#30000 and 50000
df[(df['Salary']>30000) & (df['Salary']<50000)]
```

```python
# # Sorting

# 1. df.sort_values("column_name") -> Sort by Single Column

# 2. df.sort_values("Age", ascending=True) -> Sort in Ascending Order

# 3. df.sort_values("Age", ascending=False) -> Sort in Descending Order

# 4. df.sort_values(["col1", "col2"]) -> Sort Multiple Columns

# 5. df.sort_values(["Age", "Salary"], ascending=[True, False]) -> Multiple Columns with Descending Order
```

```python
import pandas as pd

data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Riya", "Raj", "Sara", "John", "Neha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000, 52000],
    "Experience": [2, 3, 5, 4, 1, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
#sorting for ascending order
df.sort_values("Salary")
```

```python
#Two cols sort
df.sort_values(["Salary","Experience"])
```
