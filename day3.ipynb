# Operator 

1. arith -> +, -, *, /, //, %, **
2. relational -> >, <, >=, <=, ==, !=
3. assignment -> +=, -=, *=, /=, %=
4. logical -> and, or, not
5. bitwise -> &, |

```python
# Addition (+)
a = 10
b = 5
print("Addition:", a + b)

# Subtraction (-)
print("Subtraction:", a - b)

# Multiplication (*)
print("Multiplication:", a * b)

# Division (/)
print("Division:", a / b)

# Floor Division (//)
print("Floor Division:", a // b)

# Modulus (%)
print("Modulus:", a % b)

# Exponent (**)
print("Exponent:", a ** b)
```
