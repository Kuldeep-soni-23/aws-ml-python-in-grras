```python
marks = int(input("Enter The Marks: "))

if marks >= 90:
    print("A Grade")

elif marks >= 80:
    print("B Grade")

elif marks >= 70:
    print("C Grade")

elif marks >= 60:
    print("D Grade")

elif marks >= 45:
    print("E Grade")

else:
    print("FAIL")
```
